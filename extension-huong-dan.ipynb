# Hướng dẫn tạo Extension lấy dữ liệu nhà xe

Notebook này hướng dẫn từng bước tạo extension với 3 nút để lấy dữ liệu từ các API, thiết lập background để thêm header và lấy dữ liệu JSON.

## 1. Tạo Extension và Giao Diện với 3 Nút

- Tạo file `package.json` (manifest cho extension VS Code).
- Tạo file giao diện (Webview hoặc popup) với 3 nút:
  - **Lấy dữ liệu tuyến**
  - **Lấy dữ liệu tài xế**
  - **Lấy dữ liệu phương tiện**

Ví dụ giao diện HTML:
```html
<button id="btnRoute">Lấy dữ liệu tuyến</button>
<button id="btnDriver">Lấy dữ liệu tài xế</button>
<button id="btnVehicle">Lấy dữ liệu phương tiện</button>
<div id="result"></div>
```


## 2. Thiết Lập Background Script để Quản Lý Header

- Tạo file `background.js` (hoặc extension backend).
- Trong file này, thiết lập các header cần thiết cho mọi request đến các API:
  - `Authorization: Bearer <token>`
  - `cookie`
  - `referer`
  - `user-agent`
  - `accept-language`
  - `origin-request-id`
  - `origin-request-product`
  - `productname`
  - `x-data-mode`

Ví dụ (Node.js fetch):
```js
const headers = {
  'Authorization': 'Bearer <token>',
  'cookie': '<cookie>',
  'referer': '<referer>',
  'user-agent': '<user-agent>',
  'accept-language': '<accept-language>',
  'origin-request-id': '<id>',
  'origin-request-product': '<product>',
  'productname': '<productname>',
  'x-data-mode': '<mode>'
};
```


## 3. Gọi API Lấy Dữ Liệu Tuyến

- Khi nhấn nút **Lấy dữ liệu tuyến**, gửi request đến:
  - `https://nhaxe.vexere.com/api/v3/company/46249/routes`
- Sử dụng các header đã thiết lập ở background.
- Nhận dữ liệu JSON trả về.

Ví dụ fetch:
```js
fetch('https://nhaxe.vexere.com/api/v3/company/46249/routes', { headers })
  .then(res => res.json())
  .then(data => {
    // Hiển thị dữ liệu
  });
```

## 4. Gọi API Lấy Dữ Liệu Tài Xế

- Khi nhấn nút **Lấy dữ liệu tài xế**, gửi request đến:
  - `https://nhaxe.vexere.com/api/v1/driver?company_id=46249&type=2`
- Sử dụng các header đã thiết lập ở background.
- Nhận dữ liệu JSON trả về.

Ví dụ fetch:
```js
fetch('https://nhaxe.vexere.com/api/v1/driver?company_id=46249&type=2', { headers })
  .then(res => res.json())
  .then(data => {
    // Hiển thị dữ liệu
  });
```

## 5. Gọi API Lấy Dữ Liệu Phương Tiện

- Khi nhấn nút **Lấy dữ liệu phương tiện**, gửi request đến:
  - `https://nhaxe.vexere.com/api/v1/vehicle?filter[where][comp_id]=46249&filter[page]=1&filter[per_page]=500&filter[where][is_prg_status]=1`
- Sử dụng các header đã thiết lập ở background.
- Nhận dữ liệu JSON trả về.

Ví dụ fetch:
```js
fetch('https://nhaxe.vexere.com/api/v1/vehicle?filter[where][comp_id]=46249&filter[page]=1&filter[per_page]=500&filter[where][is_prg_status]=1', { headers })
  .then(res => res.json())
  .then(data => {
    // Hiển thị dữ liệu
  });
```

## 6. Hiển Thị Kết Quả JSON

- Khi nhận được dữ liệu JSON từ API, hiển thị lên giao diện extension (Webview hoặc popup).
- Có thể dùng `JSON.stringify(data, null, 2)` để định dạng đẹp.

Ví dụ:
```js
document.getElementById('result').innerText = JSON.stringify(data, null, 2);
```

---

**Lưu ý:**
- Cần thay thế các giá trị header thực tế (token, cookie, ...) cho đúng môi trường sử dụng.
- Có thể bổ sung giao diện nhập token/cookie nếu muốn động.
